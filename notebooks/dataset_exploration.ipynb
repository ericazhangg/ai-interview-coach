# Dataset Exploration

This notebook explores the real software interview dataset used by the AI Interview Coach.

It is meant for project analysis and documentation, not for running the final app.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "Software_Questions.csv"

df = pd.read_csv(DATA_PATH, encoding="latin1")
df.head()
```

## Basic Shape and Schema

```python
print("Rows:", len(df))
print("Columns:", list(df.columns))
df.info()
```

## Missing Values

```python
df.isna().sum().sort_values(ascending=False)
```

## Category Distribution

```python
category_counts = df["Category"].value_counts()
category_counts
```

```python
category_counts.plot(kind="bar", figsize=(12, 4), title="Question Count by Category")
```

## Difficulty Distribution

```python
difficulty_counts = df["Difficulty"].value_counts()
difficulty_counts
```

```python
difficulty_counts.plot(kind="bar", figsize=(6, 4), title="Question Count by Difficulty")
```

## Sample Questions and Answers

```python
df[["Question Number", "Question", "Answer", "Category", "Difficulty"]].sample(5, random_state=42)
```

## Answer Length Analysis

```python
df["answer_word_count"] = df["Answer"].fillna("").str.split().str.len()
df["question_word_count"] = df["Question"].fillna("").str.split().str.len()
df[["answer_word_count", "question_word_count"]].describe()
```

```python
df.groupby("Difficulty")[["answer_word_count", "question_word_count"]].mean().round(2)
```
